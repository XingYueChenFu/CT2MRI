```python
import os
import numpy as np
from tqdm import tqdm

# 定义路径
source_dir = "/home/featurize/data/Task1_Liu_split"
target_dir = "/home/featurize/data/Task1_Liu_split_32"

# 确保目标目录存在
os.makedirs(target_dir, exist_ok=True)

# 遍历所有.npy文件
for root, dirs, files in os.walk(source_dir):
    for file in tqdm(files):
        if file.endswith('.npy'):
            # 构建完整文件路径
            source_path = os.path.join(root, file)
            
            # 计算相对路径
            relative_path = os.path.relpath(source_path, source_dir)
            
            # 构建目标路径
            target_path = os.path.join(target_dir, relative_path)
            
            # 确保目标目录存在
            os.makedirs(os.path.dirname(target_path), exist_ok=True)
            
            # 加载数据
            data = np.load(source_path)
            
            # 检查形状是否符合预期
            if data.shape == (1, 64, 256, 256):
                # 从中间裁剪出32层
                start = (64 - 32) // 2  # 计算起始位置
                cropped_data = data[:, start:start+32, :, :]
                
                # 保存裁剪后的数据
                np.save(target_path, cropped_data)
            elif data.shape == (64, 256, 256):
                # 从中间裁剪出32层
                start = (64 - 32) // 2  # 计算起始位置
                cropped_data = data[start:start+32, :, :]
                
                # 保存裁剪后的数据
                np.save(target_path, cropped_data)
            else:
                print(f"文件 {source_path} 的形状是 {data.shape}")

print("处理完成！")
```
